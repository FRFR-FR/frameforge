```python
import torch, subprocess\nfrom google.colab import files\nfrom pathlib import Path\nprint('PyTorch:',torch.__version__,'CUDA:',torch.cuda.is_available())\nprint('Upload the 30-second watch-narration-echo.mp3 generated by FrameForge (download it from the assets provided in this chat).')\nuploaded=files.upload()\nsource_file=next(iter(uploaded))\nsubprocess.run(['ffmpeg','-y','-v','error','-i',source_file,'-t','10','-ar','24000','-ac','1','-c:a','pcm_s16le','echo-reference.wav'],check=True)\nref_text='A mechanical watch turns one wound spring into every second. Inside, a steel mainspring stores energy, not electricity. Tiny gears carry that power forward without a battery.'\nsample_text='This tiny balance wheel swings back and forth, keeping a mechanical watch ticking with remarkable precision.'\n
```
